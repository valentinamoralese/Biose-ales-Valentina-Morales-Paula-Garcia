Escalón unitario
u(n - n0) = 1 si n >= n0 ; 0 si n < n0
#
Enunciado: "Implemente una función en Python que permita generar una

función escalón unitario, definida en un intervalo n1 <= n0 <= n2."

 Se sigue la misma lógica que impseq, cambiando la condición.


```python
#FUNCIÓN ESCALON UNITARIO
def stepseq(n0, n1, n2):
    #   Genera x(n) = u(n-n0); n1 <= n <= n2
    n = np.arange(n1, n2 + 1)
    x = (n - n0) >= 0
    return [x, n]
```

Rampa

r(n - n0) = 0 si n < n0 ; (n - n0) si n >= n0

Enunciado: "Implemente una función en Python que permita generar una función rampa, definida en un intervalo [n1, n2] y una pendiente m."

```python
#FUNCIÓN RAMPA
def rampseq(n0, n1, n2, m=1):
    #   Genera x(n) = m * r(n-n0); n1 <= n <= n2
    #   ----------------------------------------------
    n = np.arange(n1, n2 + 1)
    x = np.where(n >= n0, m * (n - n0), 0)
    #np.where(condición,valor_sies_verdadero,valor_sies_falso) esto de acuerdo a las condiciones que nos plantean en el enunciado
    return [x, n]
```

Exponenciales complejas
#
x(n) = e^{(sigma + j*w0) n}, para todo n
#
sigma < 0 atenúa, sigma > 0 amplifica. Se usa np.exp.
 


```python
def cexpseq(sigma, w0, n1, n2):
    n = np.arange(n1, n2 + 1)
    x = np.exp((sigma + 1j * w0) * n)
    return [x, n]
```

Sinusoides
#
x(n) = sin(w0*n + theta0)

```python
def sinseq(w0, theta0, n1, n2):
    #   Genera x(n) = sin(w0*n + theta0); n1 <= n <= n2
    #   ----------------------------------------------
    n = np.arange(n1, n2 + 1)
    x = np.sin(w0 * n + theta0)
    #calcula el seno para cada n, pues del valor de nuestra muestra dada
    return [x, n]
```

```python
#Impulso unitario
def impseq(n0, n1, n2):
    #   Genera x(n) = delta(n-n0); n1 <= n <= n2
    #   ----------------------------------------------
    n = np.arange(n1, n2 + 1)      # Se crea el vector de muestras
    x = (n - n0) == 0
    return [x, n]
```

```python
n1 , n2 = -5,5
d1, n = impseq(-1, n1, n2)   # delta(n+1)  -> n0 = -1
d2, n = impseq(-3, n1, n2)   # delta(n+3)  -> n0 = -3
d3, n = impseq(-2, n1, n2)   # delta(n+2)  -> n0 = -2
d4, n = impseq(0, n1, n2)    # delta(n)
d5, n = impseq(0, n1, n2)    # delta(n) (segundo término del enunciado)

x1 = 3 * d1 + 5 * d2 + 3 * d3 + 3 * d4 + 1 * d5
plt.figure()
plt.stem(n, x1)
plt.title("x1(n) = 3δ(n+1) + 5δ(n+3) + 3δ(n+2) + 3δ(n) + δ(n)")
plt.xlabel("n")
plt.ylabel("x1(n)")
plt.grid(True)
plt.show()
```

1)	Escribir una función que reciba un diccionario con las notas de los alumnos de un curso y devuelva una serie con la nota mínima, la máxima, media y la desviación típica. 

```python
def estadisticas_notas(notas_dict):

    notas_series = pd.Series(notas_dict)

    estadisticas = pd.Series({
        'Mínima': notas_series.min(),
        'Máxima': notas_series.max(),
        'Media': notas_series.mean(),
        'Desviación estándar': notas_series.std()
    })

    return estadisticas
```
