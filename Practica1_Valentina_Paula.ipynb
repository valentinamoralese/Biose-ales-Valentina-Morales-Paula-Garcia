**Secuencias Discretas Elementales**

**Impulso unitario**


El impulso unitario se define como:

$$
\delta(n - n_0) = 
\begin{cases} 
1 & \text{si } n = n_0 \\
0 & \text{si } n \neq n_0
\end{cases}
$$


En Python, implementamos el impulso unitario mediante la siguiente función:




```python
def impseq(n0, n1, n2):
    #Genera x(n) = delta(n-n0); n1 <= n <= n2
    n = np.arange(n1,n2+1) # Se crea el vector de muestras
    x = (n-n0) == 0
    return [x,n]
```

**Escalón unitario**

El escalón unitario se define como:

$$
u(n - n_0) = 
\begin{cases} 
1 & \text{si } n \geq n_0 \\
0 & \text{si } n < n_0
\end{cases}
$$

p) Implemente una función en Python que permita generar una función escalón unitario, definida en un intervalo n_1≤n_0≤n_2


```python
#FUNCIÓN ESCALON UNITARIO
def stepseq(n0, n1, n2):
    #   Genera x(n) = u(n-n0); n1 <= n <= n2
    n = np.arange(n1, n2 + 1)
    x = (n - n0) >= 0
    return [x, n]
```

**Rampa**

La función rampa se define como:

$$
r(n - n_0) = 
\begin{cases} 
0 & \text{si } n < n_0 \\
n - n_0 & \text{si } n \geq n_0
\end{cases}
$$0

q)	Implemente una función en Python que permita generar una función rampa, definida en un intervalo ￼ y una pendiente m. 

```python
#FUNCIÓN RAMPA
def rampseq(n0, n1, n2, m=1):
    #   Genera x(n) = m * r(n-n0); n1 <= n <= n2
    #   ----------------------------------------------
    n = np.arange(n1, n2 + 1)
    x = np.where(n >= n0, m * (n - n0), 0)
    #np.where(condición,valor_sies_verdadero,valor_sies_falso) esto de acuerdo a las condiciones que nos plantean en el enunciado
    return [x, n]
```

**Exponenciales complejas**

Las exponenciales complejas son de la forma:

$$
x(n) = e^{(\sigma + j\omega_0)n}, \; \forall n
$$

En la que σ produce una atenuación si es negativo o una amplificación si es positivo. En Python se hace uso del comando `np.exp`.

**Sinusoides**

Estas secuencias son de la forma:

$$
x(n) = \sin(\omega_0 n + \theta_0)
$$

b. Realice un subplot con las dos secuencias, la original, y la derivada. Describa lo que observa y el porqué no es exactamente la derivada del caso continuo.

```python
##PUNTO A
##ingresamos las variables con sus respectivas funciones
ra,na = rampseq(0,0,5)
rb,nb = rampseq(5,6,11)
rc,nc = rampseq(10,12,17)

n_x= np.concatenate((na,nb,nc))
x= np.concatenate((ra,rb,rc))

##Derive la secuencia 
D = np.diff(x, prepend=x[0])
```

```python
##PUNTO B
### Subplot con la señal original y su derivada

fig, axs = plt.subplots(2, 1, sharex=True, figsize=(8, 6))
axs[0].stem(n_x, x)
axs[0].set_title("x(n): rampas concatenadas")
axs[0].set_ylabel("x(n)")
axs[0].grid(True)
 
axs[1].stem(n_x, D)
axs[1].set_title("Derivada discreta (primera diferencia): dx(n) = x(n) - x(n-1)")
axs[1].set_xlabel("n")
axs[1].set_ylabel("dx(n)")
axs[1].grid(True)
 
plt.tight_layout()
plt.show()
 
```

```python
n6 = np.arange(-20, 21)

# x(n): definida solo en [-5, 17]. Se extiende con ceros a ambos
# lados hasta cubrir todo el rango n6.
ceros_izq_xn = n_completo[0] - n6[0]      # ceros faltantes antes de -5
ceros_der_xn = n6[-1] - n_completo[-1]    # ceros faltantes después de 17
x_n_ext = np.concatenate((np.zeros(ceros_izq_xn), x_completo, np.zeros(ceros_der_xn)))

# x(n+2): es x(n) "corrida" 2 posiciones hacia la izquierda,
# por lo que existe en [-7, 15].
ceros_izq_xn2 = (n_completo[0] - 2) - n6[0]     # ceros antes de -7
ceros_der_xn2 = n6[-1] - (n_completo[-1] - 2)   # ceros después de 15
x_n2_ext = np.concatenate((np.zeros(ceros_izq_xn2), x_completo, np.zeros(ceros_der_xn2)))

x6 = 0.001 * np.exp(0.5*n6) * x_n_ext + 10 * np.sin(0.05*np.pi*n6) * x_n2_ext

plt.figure(figsize=(10, 5))
plt.stem(n6, x6)
plt.title("x6(n) = 0.001·e^(0.5n)·x(n) + 10sin(0.05πn)·x(n+2)")
plt.xlabel("n")
plt.ylabel("x6(n)")
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
def estadisticas_notas(notas_dict):

    notas_series = pd.Series(notas_dict)

    estadisticas = pd.Series({
        'Mínima': notas_series.min(),
        'Máxima': notas_series.max(),
        'Media': notas_series.mean(),
        'Desviación estándar': notas_series.std()
    })

    return estadisticas
```
